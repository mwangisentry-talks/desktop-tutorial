```python
import matplotlib.pyplot as plt

hair_color = ['Blond', 'Brown', 'Black', 'Others']
frequency = [76, 134, 160, 30]
colors = ['darkkhaki', 'red', 'yellow', 'cadetblue']  # Using lowercase color names

plt.pie(frequency, labels=hair_color, autopct='%1.1f%%', colors=colors, startangle=90)

plt.axis('equal')  # Equal aspect ratio ensures that the pie is drawn as a circle.
plt.title('Hair Color Distribution')
plt.show()
```

```python
import matplotlib.pyplot as plt

hair_color = ['Blond', 'Brown', 'Black', 'Others']
frequency = [76, 134, 160, 30]
colors = ['khaki', 'tomato', 'palegreen', 'cyan']  # Using lowercase color names
explode = [0,0,0,0.2]
plt.pie(frequency, labels=hair_color, autopct='%1.1f%%', colors=colors, startangle=90, explode=explode)


plt.axis('equal')  # Equal aspect ratio ensures that the pie is drawn as a circle.
plt.title('Hair Color Distribution')
plt.show()
```
