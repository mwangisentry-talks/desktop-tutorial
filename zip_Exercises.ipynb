```python
grades = [89,67,86,74,59,54]
students = ['Andrew', 'Tom', 'Peter', 'Doe', 'Alice','Mary']

join = zip(students,grades)

for students, grades in join:
  print(f"Students: {students}, Grades: {grades}")
```
