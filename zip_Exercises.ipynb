```python
keys = ["Name", "Age", "School", "Grade", "Gender", "Location"]
grades = [89, 67, 86]  # Three grades for three students
students = ['Andrew', 'Tom', 'Peter']

# Creating a dictionary
dict_values = dict(zip(keys, grades))  # Zip only takes two arguments

# Printing students and grades
for student, grade in zip(students, grades):
    print(f"Student: {student}, Grade: {grade}")

# Printing the dictionary
print(dict_values)
```
