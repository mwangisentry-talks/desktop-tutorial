```python
import statistics

data_grades = [89,65,74,81,76,84]
data_previous_score = [78,61,65,71,73,73]

corr_coefficient = statistics.correlation(data_grades,data_previous_score)

q1_data_grades = statistics.quantiles(data_grades, n=4)[0]
q3_data_grades = statistics.quantiles(data_grades, n=4)[-1]


q1_data_previous_score = statistics.quantiles(data_previous_score, n=4)[0]
q3_data_previous_score = statistics.quantiles(data_previous_score, n=4)[-1]



print("Correlation Coefficient is : ", corr_coefficient)

print("First Quartile:", q1_data_grades)
print("Third Quartile:", q3_data_grades)

print("First Quartile:", q1_data_previous_score)
print("Third Quartile:", q3_data_previous_score)
```
