```python
import numpy as np

matrix3 = np.array([['Garnacho', 'Mainoo'], ['Rooney', 'Nani'], ['PowerBi', 'Spark']])

# Make a copy of the array
matrix3_copy = matrix3.copy()

# Save the matrix to a binary file
np.save('matrix3.npy', matrix3)

# Save the copy to a CSV file
np.savetxt("matrix3_copy.csv", matrix3_copy, fmt='%s', delimiter=",")

print('---+--Copy matrix---+--\n')
print(matrix3_copy)

print('---+--Original matrix---+--\n')
print(matrix3)

# Check the data type of the array
print("\nData Type of the Array:")
print(matrix3.dtype)
new_array = np.load("matrix3.npy")
print(new_array)

print('\n')
print('Load matrix3_copy.csv file read lines\n')
matricFile = open("matrix3_copy.csv", "r")
for i in matricFile:
    print(i)
   
matricFile.close()
```
