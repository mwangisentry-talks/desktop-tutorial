```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

corpus = ['i earn ten bob an hour at the deli', 'can do bug bounty haacking']

dataframe = pd.DataFrame({'Text': corpus})
count_v = CountVectorizer()

x = count_v.fit_transform(dataframe.Text).toarray()

print(x)
print(count_v.vocabulary)
```

```python
new_txt = ['io etrn are you free ruppee for a monnth meeting chitcchting anytime tomorrow neet']
df_new = pd.DataFrame({'new_txt':new_txt})
y = count_v.transform(df_new.new_txt).toarray()
print(y)
```
